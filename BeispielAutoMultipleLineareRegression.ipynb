```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df = pd.DataFrame({
    "Jahr": [2018,2019,2020,2021,2022,2023,2024],
    "Kilometer": [120000,100000,80000,60000,50000,30000,10000],
    "PS": [150,150,160,170,180,190,200],
    "Preis": [21000,23000,25000,27000,29000,31000,33000]
})
```

```python
%matplotlib inline
plt.scatter(df.Jahr,df.Preis)
```

```python
plt.xlabel('Jahr')
plt.ylabel("Preis")
plt.scatter(df.Jahr,df.Preis, color="red", marker="+")
```

```python
X = df[["Jahr", "Kilometer", "PS"]]
y = df["Preis"]
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
print("Steigung w:", model.coef_[0])
print("model.coef_:", model.coef_)
print("Achsenabschnitt b:", model.intercept_)
```

```python
vorhersage = model.predict(pd.DataFrame({
    "Jahr": [2025],
    "Kilometer": [5000],
    "PS": [210]
}))

print(vorhersage[0])
```

```python
model = LinearRegression()
model.fit(X, y)

print(model.coef_)
print(model.intercept_)
df["Vorhersage"] = model.predict(df[["Jahr", "Kilometer", "PS"]])

print(df[["Preis", "Vorhersage"]])

model.score(X, y)
```

```python
%matplotlib inline
plt.plot(
    df["Jahr"],
    model.predict(df[["Jahr", "Kilometer", "PS"]]),
    color="blue"
)
```
